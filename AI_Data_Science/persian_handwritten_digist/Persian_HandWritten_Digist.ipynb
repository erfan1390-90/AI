```python
import kagglehub
import os
path = kagglehub.dataset_download("amirmahdiabbootalebi/persian-handwritten-digits")
print("Path to dataset files:", path)
print("Contents of the dataset directory:", os.listdir(path))
```

```python
import os
import tensorflow as tf
import numpy as np
from PIL import Image
```

```python
train_dir = os.path.join(path, "Train")
val_dir = os.path.join(path, "Test")
train_data = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    image_size=(28, 28),
    batch_size=32,
    shuffle=True,
    color_mode='grayscale'
)
val_data = tf.keras.utils.image_dataset_from_directory(
    val_dir,
    image_size=(28, 28),
    batch_size=32,
    shuffle=False,
    color_mode='grayscale'
)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Rescaling(1./ 255, input_shape=(28, 28, 1)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation="softmax")
])
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.fit(train_data, epochs=25)
```

```python
model.save("Persian_Digist.keras")
```

```python
model.evaluate(val_data)
```
